```python
class computer():        
     
    def __init__(self,cpu,ram): # cpu & ram as argument
        self.cpu = cpu #hear self is the object & we need to refer to the object
        self.ram = ram 
 
    def config(self):     
        print('config is', self.cpu, self.ram) 

com1 = computer('i5', 8)
com2 = computer('i7', 16)


com1.config() 
com2.config()

#print(id(com1))
#print(id(com2))
```

```python
class computer():
    def __init__(self,cpu,ram,price):
        self.cpu=cpu
        self.ram = ram
        self.price = price

    def config(self):
        print('config is',self.cpu, self.ram, self.price)

com1= computer('i5',8,50000)
com2= computer('i7',16,70000)

com1.config()
com2.config()
```

```python
class computer():
    def __init__(self,cpu,ram,price,company):
        self.cpu = cpu
        self.ram = ram
        self.price = price
        self.company= company

    def config(self):
        print('config is', self.cpu,self.ram,self.price,self.company)

com1= computer('i5',8,50000,'len')
com2= computer('i7',16,70000,'hp')

com1.config()
com2.config()
```
